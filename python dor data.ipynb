```python
# Converting the currency columns from text to int 
Currency_Columns=[
    "Unit Price",
    "Unit Cost","Revenue","Unit Cost","Profit"
]
for col in Currency_Columns:
    df[col]=(df[col]
    .astype(str)
    .str.replace("$", "",regex=False)
    .str.replace(",", "",regex=False)
    .astype(float)
    )  
```

```python
region_revenue=df.groupby("Region")["Revenue"].sum().sort_values(ascending=False)
plt.figure(figsize=(8,5))
colors=plt.cm.Set2(range(len(region_revenue)))
bars=plt.bar(region_revenue.index,region_revenue.values,color=colors)
plt.title("Revenue By Region")
plt.xlabel("Regions")
plt.ylabel("Revenue")
plt.bar_label(bars)
plt.xticks(rotation=45)
```

```python
region_profit=df.groupby("Region")["Profit"].sum().sort_values(ascending=False)
plt.figure(figsize=(8,5))
colors=plt.cm.Set2(range(len(region_profit)))
bars=plt.bar(region_profit.index,region_profit.values,color=colors)
plt.title("Profit Per Region")
plt.xlabel("Regions")
plt.ylabel("Revenue")
plt.bar_label(bars)
```

```python
category_revenue=df.groupby("Category")["Revenue"].sum().sort_values(ascending=False)
colors=plt.cm.Set2(range(len(category_revenue)))
bars=plt.bar(category_revenue.index,category_revenue.values,color=colors)
plt.title("Category Wise Revenue")
plt.xlabel("categories")
plt.ylabel("Revenue")
plt.bar_label(bars)
plt.xticks(rotation=45)
```

```python
category_profit=df.groupby("Category")["Profit"].sum().sort_values(ascending=False)
plt.bar(category_profit.index,category_profit.values)
plt.title("Profit By Category")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.figure(figsize=(5,5))
```

```python
product_profit=df.groupby("Product")["Profit"].sum().sort_values()
plt.barh(product_profit.index,product_profit.values)
plt.title("Profit By Product")
plt.xlabel("Profit")
plt.ylabel("Product")
```

```python
daily_revenue=df.groupby("Order Date")["Revenue"].sum()
plt.figure(figsize=(16,6))
plt.plot(daily_revenue.index,daily_revenue.values,marker="o")
plt.title("Revenue By Date")
plt.xlabel("Order Dates")
plt.ylabel("Revenue")
plt.xticks(rotation=40)
```

```python
channel_revenue=df.groupby("Channel")["Revenue"].sum()
colors=plt.cm.Set2(range(len(channel_revenue)))
plt.pie(channel_revenue.values,
        labels=channel_revenue.index,
        autopct="%1.1f%%",
        colors=colors)
plt.title("Channel Revenue")
```

```python
category_revenue=df.groupby("Category")["Revenue"].sum()
plt.figure(figsize=(7,7))
color=plt.cm.Set2(range(len(category_revenue)))
plt.pie(
    category_revenue.values,
    labels=category_revenue.index,
    autopct="1%.1f%%",
    colors=color,
    wedgeprops={"width":0.4}
)
plt.title("Revenue by category")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Revenue"],bins=10,color="Yellow")
plt.title("Revenue Dustribution")
plt.xlabel("Revenue")
plt.ylabel("frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.kdeplot(df["Profit"],fill=True)
plt.title("Profit Distributuion")
plt.xlabel("Profit")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["Profit"],color="Red")
plt.title("Profit Outlier Analysis")
plt.xlabel("Profit")
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Revenue"],
        df["Profit"],
        alpha=0.3 , #transperancy reveals density instead of a solid blob
        s=15,
        color="#4C72B0",
        edgecolors="None"
                )
plt.title("Revenue VS Profit")
plt.xlabel("Revenue")
plt.ylabel("Profit")
plt.tight_layout()
```

```python
plt.figure(figsize=(8,5))
bars=sns.countplot(
    data=df,
    x="Region")
plt.title("Number of Orders Per Region")
plt.xlabel("Region")
plt.ylabel("Number of Orderss")
plt.bar_label(bars.containers[0])
```

```python
plt.figure(figsize=(8,5))
bars=sns.countplot(
    data=df,
    x="Channel"
)
plt.title("Numbers of Orders By Channel")
plt.xlabel("Channel")
plt.ylabel("Number of orders")
plt.bar_label(  bars.containers[0])
```

```python
pivot_revenue=pd.pivot_table(
    df,
    values="Revenue",
    index="Region",
    columns="Category",
    aggfunc="sum"
)
plt.figure(figsize=(10,7))
sns.heatmap(pivot_revenue,
            annot=True,
            fmt=".0f")
plt.title("Revenue By Region and Category")
plt.show()
```

```python
region_data=df.groupby("Region")[["Revenue","Profit"]].sum()
region_data.plot(
    kind="bar",figsize=(10,9)
)
plt.title("Revenue VS Profit ByvRegion")
plt.xlabel("Region")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.show()
```

```python
daily_revenue=df.groupby("Order Date")["Revenue"].sum()
plt.figure(figsize=(12,5))
plt.fill_between(
    daily_revenue.index,
    daily_revenue.values,
    color="violet"
)
plt.title("Revenue Trend")
plt.xlabel("Order Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.violinplot(
    data=df,
    x="Category",
    y="Profit"
)
plt.title("Profit Distribution By Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.show()
```
